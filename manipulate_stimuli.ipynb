```python
import os
import argparse
import numpy as np
from PIL import Image, ImageFilter, ImageEnhance
```

```python
# Functions to define the perturbations

def add_noise(image, noise_std=25):
    arr = np.array(image).astype(np.float32)
    noise = np.random.normal(0, noise_std, arr.shape)
    noisy = arr + noise
    noisy = np.clip(noisy, 0, 255).astype(np.uint8)
    return Image.fromarray(noisy)


def alter_image(image, alteration, strength):
    if alteration == "blur":
        return image.filter(ImageFilter.GaussianBlur(radius=strength))

    elif alteration == "noise":
        return add_noise(image, noise_std=strength)

    elif alteration == "contrast":
        enhancer = ImageEnhance.Contrast(image)
        return enhancer.enhance(strength)

    elif alteration == "brightness":
        enhancer = ImageEnhance.Brightness(image)
        return enhancer.enhance(strength)

    elif alteration == "grayscale":
        return image.convert("L").convert("RGB")

    else:
        raise ValueError(f"Unknown alteration: {alteration}")
```

```python
def process_folder(input_dir, output_dir, alterations):
    os.makedirs(output_dir, exist_ok=True)

    for fname in os.listdir(input_dir):
        if not fname.lower().endswith(".png"):
            continue

        in_path = os.path.join(input_dir, fname)
        image = Image.open(in_path).convert("RGB")

        base = os.path.splitext(fname)[0]

        for alteration, strength in alterations:
            altered = alter_image(image, alteration, strength)

            out_name = f"{base}.png"
            os.makedirs(os.path.join(output_dir, f"{alteration}_{strength}"), exist_ok=True)
            out_path = os.path.join(output_dir, f"{alteration}_{strength}", out_name)

            altered.save(out_path)

    print(f"Saved altered images to: {output_dir}")
```

```python
input_dir = "./images/obj_2AFC"
```

```python
output_dir = "./images_perturbed/obj_2AFC"
```

```python
#Define the perturbations
alterations = [
      ("blur", 2),
      ("blur", 5),
      ("noise", 15),
      ("noise", 40),
      ("contrast", 0.5),
      ("contrast", 1.5),
      ("brightness", 0.6),
      ("brightness", 1.4),
      ("grayscale", 1),
  ]
```

```python
process_folder(input_dir, output_dir, alterations)
```
