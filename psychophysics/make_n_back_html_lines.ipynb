```python
import pandas as pd
import json
import random
```

```python
# =========================
# STUDENTS: EDIT THIS PART
# =========================
meta_file = "examples_metadata_files/metadata_nback.csv"
random_seed = 1

# Default N-Back value
# N = 5 means: repeated images appear again 5 trials later
N = 5
# =========================
```

```python
random.seed(random_seed)

df = pd.read_csv(meta_file)

required_columns = ["image_file", "image_idx", "repeated"]
for col in required_columns:
    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")

df["image_idx"] = df["image_idx"].astype(int)
df["repeated"] = df["repeated"].astype(int)

if not df["repeated"].isin([0, 1]).all():
    raise ValueError("The repeated column must contain only 0 or 1.")

# Randomize the order of the first presentation of all images
df = df.sample(frac=1, random_state=random_seed).reset_index(drop=True)

sequence = []
correct = []

for _, row in df.iterrows():
    sequence.append(int(row["image_idx"]))
    correct.append(0)  # first time = Novel

# Add repeated images N trials after their first presentation
repeat_images = df[df["repeated"] == 1]["image_idx"].astype(int).tolist()

for img_idx in repeat_images:
    first_pos = sequence.index(img_idx)
    repeat_pos = first_pos + N

    if repeat_pos > len(sequence):
        sequence.append(img_idx)
        correct.append(1)
    else:
        sequence.insert(repeat_pos, img_idx)
        correct.insert(repeat_pos, 1)

print("// Copy these lines into the N-Back HTML file")
print("let val = " + json.dumps(sequence) + ";")
print("let correct = " + json.dumps(correct) + ";")
```
