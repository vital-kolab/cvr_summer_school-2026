```python
import pandas as pd
import json
import random
```

```python
# =========================
# STUDENTS: EDIT THIS PART
# =========================
meta_file = "examples_metadata_files/metadata_rating.csv"
random_seed = 1
# =========================
```

```python


random.seed(random_seed)

df = pd.read_csv(meta_file)

required_columns = ["image_file", "image_idx"]
for col in required_columns:
    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")

# Randomize image order
df = df.sample(frac=1, random_state=random_seed).reset_index(drop=True)

val = df["image_idx"].astype(int).tolist()

print("Copy this line into the Rating HTML file:")
print("let val = " + json.dumps(val) + ";")
```
