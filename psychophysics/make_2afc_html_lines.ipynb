```python
import pandas as pd
import json
import random
```

```python
# =========================
# STUDENTS: EDIT THIS PART
# =========================
meta_file = "examples_metadata_files/metadata_2afc.csv"
random_seed = 1

# Class list used by the task
# The numbers in correct_choice refer to this list.
lb = ['bear', 'elephant', 'person', 'car', 'dog',
      'apple', 'chair', 'plane', 'bird', 'zebra']
# =========================
```

```python
random.seed(random_seed)

df = pd.read_csv(meta_file)

required_columns = ["image_file", "image_idx", "correct_choice"]
for col in required_columns:
    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")

df["image_idx"] = df["image_idx"].astype(int)
df["correct_choice"] = df["correct_choice"].astype(int)

all_classes = list(range(len(lb)))

# Check that all correct answers are valid
for correct in df["correct_choice"]:
    if correct not in all_classes:
        raise ValueError(
            f"Invalid correct_choice: {correct}. "
            f"It must be between 0 and {len(lb) - 1}."
        )

# Randomize image order
df = df.sample(frac=1, random_state=random_seed).reset_index(drop=True)

val = []
correct_choice = []
ch1 = []
ch2 = []

for _, row in df.iterrows():
    image_idx = int(row["image_idx"])
    correct = int(row["correct_choice"])

    # Choose one wrong class
    wrong_options = [c for c in all_classes if c != correct]
    wrong = random.choice(wrong_options)

    # Randomize whether the correct answer is on the left or right
    if random.random() < 0.5:
        left_choice = correct
        right_choice = wrong
    else:
        left_choice = wrong
        right_choice = correct

    val.append(image_idx)
    correct_choice.append(correct)
    ch1.append(left_choice)
    ch2.append(right_choice)

print("// Copy these lines into the 2AFC HTML file")
print("let val = " + json.dumps(val) + ";")
print("let ch1 = " + json.dumps(ch1) + ";")
print("let ch2 = " + json.dumps(ch2) + ";")
print("let correct_choice = " + json.dumps(correct_choice) + ";")
```
